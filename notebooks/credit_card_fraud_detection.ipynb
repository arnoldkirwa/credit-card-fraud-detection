# 1. Business Understanding

## 1.1 Project Overview

Credit card fraud is a significant challenge in the financial industry, resulting in financial losses for cardholders, banks, and other financial institutions. As digital payment transactions increase, identifying fraudulent activities accurately and promptly becomes increasingly important.

This project focuses on developing a machine learning-based credit card fraud detection system that classifies transactions as legitimate or potentially fraudulent using historical transaction data. The project will follow the Cross-Industry Standard Process for Data Mining (CRISP-DM) to guide business understanding, data exploration, data preparation, model development, evaluation, and deployment.

The final deliverable will include an evaluated machine learning model and a Streamlit web application that demonstrates how transaction predictions can be generated.

## 1.2 Business Problem

Financial institutions process large volumes of credit card transactions, most of which are legitimate. However, a small proportion may involve fraudulent activity. Because fraudulent transactions are relatively rare, detecting them is a highly imbalanced classification problem.

A model that predicts every transaction as legitimate could achieve high overall accuracy while failing to identify actual fraud. Conversely, a model that flags too many legitimate transactions may create unnecessary investigations, inconvenience customers, and increase operational costs.

The business problem is therefore to develop a reliable classification approach that identifies fraudulent transactions while balancing fraud detection, false alarms, and operational efficiency.

## 1.3 Problem Statement

The objective of this project is to investigate whether machine learning can effectively distinguish fraudulent credit card transactions from legitimate transactions using historical transaction features.

The project will analyze the dataset, address relevant data preparation challenges, train and compare classification models, and evaluate their ability to detect fraud. Model selection will prioritize appropriate fraud detection performance rather than accuracy alone.

The selected model will subsequently be prepared for demonstration through a Streamlit application. Its predictions will support an illustrative fraud detection workflow and will not be treated as definitive proof of fraud.

### 2.9.4 Outlier Detection Using the IQR Method

The Interquartile Range (IQR) method will be used to identify potential outliers in transaction amounts. Values below the lower bound or above the upper bound will be flagged for investigation.

Identifying a potential outlier does not mean the transaction is fraudulent or incorrect. No records will be removed at this stage.

```python
# Step 2.9.4: Detect potential outliers in transaction amounts

Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Amount"] < lower_bound) |
    (df["Amount"] > upper_bound)
]

print(f"Q1: {Q1:.2f}")
print(f"Q3: {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Lower bound: {lower_bound:.2f}")
print(f"Upper bound: {upper_bound:.2f}")
print(f"Potential outliers: {len(outliers):,}")
print(
    f"Percentage of potential outliers: "
    f"{len(outliers) / len(df) * 100:.2f}%"
)
```

### 2.9.5 Interpretation of Outlier Detection Results

The IQR method identified 31,904 potential outliers in transaction amounts, representing 11.20% of the dataset.

The lower bound is -101.75, while the upper bound is 184.51. Since transaction amounts are non-negative, the potential outliers are amounts above the upper bound.

These transactions are not necessarily errors or fraudulent activities. They will be retained for further investigation during the Data Preparation phase.

# 4. Modeling

## 4.1 Introduction

The modeling phase focuses on developing machine-learning models to identify potentially fraudulent credit card transactions. The prepared training dataset will be used to train the models, while the testing dataset will be reserved for final evaluation.

Because fraudulent transactions represent a very small proportion of the dataset, model selection will prioritize the ability to detect fraud while controlling false positives. ## 4.2 Import Modeling Libraries

This step imports the tools required to build machine-learning pipelines, train classification models, and evaluate their performance. Pipelines will allow preprocessing and model training to be managed together, helping prevent data leakage.

```python
# Step 4.2: Import modeling libraries

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

print("Modeling libraries imported successfully.")
```

## 4.3 Build the Logistic Regression Pipeline

Logistic Regression is used as the baseline classification model for detecting fraudulent credit card transactions.

A machine-learning pipeline will combine feature scaling and Logistic Regression into a single workflow. StandardScaler will standardize numerical features, while `class_weight="balanced"` will assign greater weight to the minority fraud class during training.

The pipeline will be fitted using the training data only. The testing data will remain unseen during training and will be used later for evaluation.

```python
# Step 4.3: Build the Logistic Regression pipeline

logistic_regression_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

print("Logistic Regression pipeline created successfully.")
print("\nPipeline steps:")
print(logistic_regression_pipeline)
```

### 4.4 Train the Logistic Regression Model

In this step, the Logistic Regression pipeline is fitted using the training dataset. The StandardScaler learns the feature scaling parameters from the training data, and the Logistic Regression classifier learns patterns that distinguish legitimate transactions from fraudulent transactions.

Class weighting is enabled to account for the severe class imbalance. The testing dataset will not be used during training.

```python
# Step 4.4: Train the Logistic Regression model

logistic_regression_pipeline.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
print("Training records:", X_train.shape[0])
print("Number of features:", X_train.shape[1])
```

### 4.5 Generate Predictions Using Logistic Regression

The trained Logistic Regression pipeline will predict whether transactions in the testing dataset are legitimate or fraudulent. The test dataset was kept separate during training to evaluate how well the model generalizes to unseen transactions.

The model will generate both predicted class labels and fraud probabilities. These outputs will be used in the subsequent evaluation steps.

```python
# Step 4.5: Generate predictions on the test set

# Predict transaction classes
y_pred_lr = logistic_regression_pipeline.predict(X_test)

# Predict the probability of fraud (Class 1)
y_proba_lr = logistic_regression_pipeline.predict_proba(X_test)[:, 1]

print("Predictions generated successfully.")
print("Number of test transactions:", len(y_pred_lr))
print("Predicted legitimate transactions:", (y_pred_lr == 0).sum())
print("Predicted fraudulent transactions:", (y_pred_lr == 1).sum())

print("\nFirst 10 predicted classes:")
print(y_pred_lr[:10])

print("\nFirst 10 fraud probabilities:")
print(y_proba_lr[:10])
```

### 4.6 Evaluate Logistic Regression Performance

This step evaluates the Logistic Regression model using the predictions generated from the test dataset. Precision, recall, F1-score, the confusion matrix, and Average Precision (PR-AUC-related) will be used to assess fraud detection performance.

These metrics are more informative than accuracy alone because fraudulent transactions represent a very small proportion of the dataset. The results will help identify the trade-off between detecting fraud and incorrectly flagging legitimate transactions.

```python
# Step 4.6: Evaluate Logistic Regression

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

# Calculate evaluation metrics
precision_lr = precision_score(y_test, y_pred_lr, zero_division=0)
recall_lr = recall_score(y_test, y_pred_lr, zero_division=0)
f1_lr = f1_score(y_test, y_pred_lr, zero_division=0)
pr_auc_lr = average_precision_score(y_test, y_proba_lr)

print("LOGISTIC REGRESSION EVALUATION")
print("-" * 40)

print(f"Precision: {precision_lr:.4f}")
print(f"Recall:    {recall_lr:.4f}")
print(f"F1-score:  {f1_lr:.4f}")
print(f"Average Precision (PR-AUC summary): {pr_auc_lr:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred_lr,
    target_names=["Legitimate", "Fraud"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

#### 4.6.1 Interpretation of Logistic Regression Results

The Logistic Regression baseline achieved a precision of 5.64%, a recall of 87.37%, an F1-score of 10.59%, and an Average Precision score of 0.6719.

The model correctly identified 83 of the 95 fraudulent transactions in the test set, missing 12 fraud cases. However, it incorrectly flagged 1,389 legitimate transactions as fraudulent, resulting in low precision.

These results demonstrate a trade-off between detecting fraudulent transactions and minimizing false alarms. The recall indicates that the model detected most fraud cases, but its low precision suggests that many flagged transactions were legitimate. The Average Precision score provides an additional measure of the model's ability to rank fraudulent transactions above legitimate ones.

The Logistic Regression model will serve as the baseline for comparison with the other classification models. Its performance does not yet establish it as the best model for deployment.

## 4.7 Build the Decision Tree Pipeline

A Decision Tree is a supervised machine-learning algorithm that learns decision rules to classify transactions as legitimate or fraudulent. Unlike Logistic Regression, it can capture nonlinear relationships between input features and the target variable.

Feature scaling is generally unnecessary for Decision Trees. The model will use class weighting to help account for the imbalance between legitimate and fraudulent transactions.

```python
# Step 4.7: Build the Decision Tree pipeline

from sklearn.tree import DecisionTreeClassifier

decision_tree_pipeline = Pipeline([
    ("model", DecisionTreeClassifier(
        class_weight="balanced",
        random_state=42,
        max_depth=10,
        min_samples_leaf=5
    ))
])

print("Decision Tree pipeline created successfully.")
print("\nPipeline steps:")
print(decision_tree_pipeline)
```

### 4.8 Train the Decision Tree Model

The Decision Tree pipeline will be fitted using the training dataset. Class weighting will help account for the imbalance between legitimate and fraudulent transactions.

After training, performance on both the training and testing sets will be compared to assess generalization and identify possible overfitting or underfitting.

```python
# Step 4.8: Train the Decision Tree model

decision_tree_pipeline.fit(X_train, y_train)

print("Decision Tree model trained successfully.")
print("Training records:", X_train.shape[0])
print("Number of features:", X_train.shape[1])
```

### 4.9 Assess Decision Tree Generalization

This step compares the Decision Tree model's performance on the training and testing datasets. A large performance gap may indicate overfitting, while poor performance on both datasets may indicate underfitting.

Because the dataset is highly imbalanced, the comparison will focus on precision, recall, F1-score, and Average Precision rather than accuracy alone.

```python
# Step 4.9: Compare Decision Tree training and testing performance

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

# Predictions on training data
y_train_pred_dt = decision_tree_pipeline.predict(X_train)
y_train_proba_dt = decision_tree_pipeline.predict_proba(X_train)[:, 1]

# Predictions on testing data
y_test_pred_dt = decision_tree_pipeline.predict(X_test)
y_test_proba_dt = decision_tree_pipeline.predict_proba(X_test)[:, 1]

# Calculate metrics for both datasets
results_dt = {
    "Dataset": ["Training", "Testing"],
    "Precision": [
        precision_score(y_train, y_train_pred_dt, zero_division=0),
        precision_score(y_test, y_test_pred_dt, zero_division=0)
    ],
    "Recall": [
        recall_score(y_train, y_train_pred_dt, zero_division=0),
        recall_score(y_test, y_test_pred_dt, zero_division=0)
    ],
    "F1-score": [
        f1_score(y_train, y_train_pred_dt, zero_division=0),
        f1_score(y_test, y_test_pred_dt, zero_division=0)
    ],
    "Average Precision": [
        average_precision_score(y_train, y_train_proba_dt),
        average_precision_score(y_test, y_test_proba_dt)
    ]
}

results_dt_df = pd.DataFrame(results_dt).set_index("Dataset")
display(results_dt_df.round(4))
```

#### 4.9.1 Interpretation of Decision Tree Generalization

### 4.16 Build the XGBoost Pipeline

XGBoost is evaluated as the fourth candidate model. Because the dataset is highly imbalanced, the model will use a class-imbalance weight calculated from the training labels. We will compare its precision, recall, F1-score, Average Precision, and training-versus-testing performance against the previous models.

### 4.17 Build the XGBoost Pipeline

XGBoost uses gradient boosting to build trees sequentially. The scale_pos_weight parameter will account for class imbalance using the ratio of legitimate to fraudulent transactions in the training set. The model will be evaluated on both training and testing data to assess generalization and possible overfitting.

```python
# Step 4.17: Build the XGBoost pipeline

from xgboost import XGBClassifier
from sklearn.pipeline import Pipeline

# Calculate class imbalance using training labels only
scale_pos_weight = (y_train == 0).sum() / (y_train == 1).sum()

xgboost_pipeline = Pipeline([
    ("model", XGBClassifier(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        objective="binary:logistic",
        eval_metric="aucpr",
        random_state=42,
        n_jobs=-1
    ))
])

print("XGBoost pipeline created successfully.")
print("Scale positive weight:", round(scale_pos_weight, 2))
```

```python
# Step 4.18: Train the XGBoost model

xgboost_pipeline.fit(X_train, y_train)

print("XGBoost model trained successfully.")
print("Training records:", X_train.shape[0])
print("Number of features:", X_train.shape[1])
```

```python
# Step 4.19: Generate XGBoost predictions

# Predictions on training data
y_train_pred_xgb = xgboost_pipeline.predict(X_train)
y_train_proba_xgb = xgboost_pipeline.predict_proba(X_train)[:, 1]

# Predictions on testing data
y_test_pred_xgb = xgboost_pipeline.predict(X_test)
y_test_proba_xgb = xgboost_pipeline.predict_proba(X_test)[:, 1]

print("XGBoost predictions generated successfully.")
print("Training transactions:", len(y_train_pred_xgb))
print("Testing transactions:", len(y_test_pred_xgb))

print("\nTest-set prediction counts:")
print("Legitimate predictions:", (y_test_pred_xgb == 0).sum())
print("Fraud predictions:", (y_test_pred_xgb == 1).sum())
```

### 4.23 Classification Threshold Analysis

The default classification threshold may not be optimal for fraud detection. This step examines how different thresholds affect precision, recall, and F1-score using Random Forest predicted probabilities. Threshold selection will use training data to avoid choosing a threshold based on the held-out test labels.

```python
# Step 4.23: Compare Random Forest thresholds using training data

import numpy as np
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score

threshold_results = []

for threshold in [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]:
    train_pred_threshold = (
        y_train_proba_rf >= threshold
    ).astype(int)

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_train, train_pred_threshold, zero_division=0
        ),
        "Recall": recall_score(
            y_train, train_pred_threshold, zero_division=0
        ),
        "F1-score": f1_score(
            y_train, train_pred_threshold, zero_division=0
        )
    })

threshold_results_df = pd.DataFrame(threshold_results)

display(threshold_results_df.round(4))
```

### 4.24 Create a Validation Set for Threshold Selection

A validation set will help us choose a suitable classification threshold without using the final test set. We will preserve the class distribution because fraudulent transactions are extremely rare.

```python
# Step 4.24: Create training and validation sets

from sklearn.model_selection import train_test_split

X_train_model, X_val, y_train_model, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

print("Training subset shape:", X_train_model.shape)
print("Validation set shape:", X_val.shape)

print("\nFraud cases in training subset:")
print(y_train_model.value_counts())

print("\nFraud cases in validation set:")
print(y_val.value_counts())
```

### 4.25 Retrain Random Forest Using the Training Subset

The Random Forest is retrained on the reduced training subset so that the validation set can be used independently for threshold selection. The final test set remains reserved for the final model evaluation.


```python
# Step 4.25: Retrain Random Forest

from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

random_forest_validation_pipeline = Pipeline([
    ("model", RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced",
        max_depth=12,
        min_samples_leaf=5,
        random_state=42,
        n_jobs=-1
    ))
])

random_forest_validation_pipeline.fit(
    X_train_model,
    y_train_model
)

print("Random Forest retrained successfully.")
print("Training records:", len(X_train_model))
print("Validation records:", len(X_val))
print("Fraud cases in validation set:", int(y_val.sum()))
```

### 4.26 Generate Validation Probabilities

The trained Random Forest will estimate the probability that each validation transaction is fraudulent. These probabilities will allow us to compare classification thresholds and evaluate the trade-off between precision and recall without using the final test set.

```python
# Step 4.26: Generate validation probabilities

y_val_proba_rf = random_forest_validation_pipeline.predict_proba(X_val)[:, 1]

print("Validation probabilities generated successfully.")
print("Number of predictions:", len(y_val_proba_rf))
print("Minimum predicted fraud probability:", round(y_val_proba_rf.min(), 4))
print("Maximum predicted fraud probability:", round(y_val_proba_rf.max(), 4))
```

### 4.27 Validation Threshold Comparison

Different classification thresholds affect fraud detection performance. We will compare precision, recall, and F1-score on the validation set to identify a suitable threshold before evaluating the final model on the held-out test set.

```python
# Step 4.27: Compare thresholds using validation data

import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score

validation_threshold_results = []

for threshold in [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]:
    y_val_pred_threshold = (
        y_val_proba_rf >= threshold
    ).astype(int)

    validation_threshold_results.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_val, y_val_pred_threshold, zero_division=0
        ),
        "Recall": recall_score(
            y_val, y_val_pred_threshold, zero_division=0
        ),
        "F1-score": f1_score(
            y_val, y_val_pred_threshold, zero_division=0
        ),
        "Fraud Flagged": int(y_val_pred_threshold.sum())
    })

validation_threshold_df = pd.DataFrame(
    validation_threshold_results
)

display(
    validation_threshold_df.round(4)
)
```

### 4.28 Final Test Evaluation at the Selected Threshold

The threshold of 0.8 was selected using validation data. We will now evaluate the Random Forest on the untouched test set to estimate how well it generalizes to unseen transactions. The test results will not be used to adjust the threshold.

```python
# Step 4.28: Evaluate selected threshold on the test set

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score
)

selected_threshold = 0.8

y_test_proba_rf_tuned = (
    random_forest_validation_pipeline.predict_proba(X_test)[:, 1]
)

y_test_pred_rf_tuned = (
    y_test_proba_rf_tuned >= selected_threshold
).astype(int)

print("Selected threshold:", selected_threshold)

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_pred_rf_tuned))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_test_pred_rf_tuned,
    target_names=["Legitimate", "Fraud"],
    zero_division=0
))

print("Precision:", round(
    precision_score(y_test, y_test_pred_rf_tuned, zero_division=0), 4
))
print("Recall:", round(
    recall_score(y_test, y_test_pred_rf_tuned, zero_division=0), 4
))
print("F1-score:", round(
    f1_score(y_test, y_test_pred_rf_tuned, zero_division=0), 4
))
print("Average Precision:", round(
    average_precision_score(y_test, y_test_proba_rf_tuned), 4
))
```

```python
# Step 5.7: Compare validation thresholds by illustrative cost

import pandas as pd

cost_per_missed_fraud = 100
cost_per_false_alarm = 5

cost_results = []

for threshold in [0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]:
    val_predictions = (y_val_proba_rf >= threshold).astype(int)

    missed_fraud = int(
        ((y_val == 1) & (val_predictions == 0)).sum()
    )

    false_alarms = int(
        ((y_val == 0) & (val_predictions == 1)).sum()
    )

    estimated_cost = (
        missed_fraud * cost_per_missed_fraud
        + false_alarms * cost_per_false_alarm
    )

    cost_results.append({
        "Threshold": threshold,
        "Missed Fraud": missed_fraud,
        "False Alarms": false_alarms,
        "Estimated Cost": estimated_cost
    })

validation_cost_df = pd.DataFrame(cost_results)

display(
    validation_cost_df
    .sort_values("Estimated Cost")
    .reset_index(drop=True)
)
```

### 5.7 Business Cost Analysis Findings

The validation-set cost comparison identified a classification threshold of 0.3 as the lowest-cost option among the nine thresholds evaluated. Under the illustrative assumptions of 100 cost units per missed fraud case and 5 units per false alarm, this threshold produced an estimated cost of 1,085 units, with 9 missed fraud cases and 37 false alarms.

The result demonstrates that the threshold with the highest precision or F1-score is not necessarily the threshold with the lowest business cost. Threshold selection should reflect the relative costs of missed fraud and unnecessary investigations.

These estimates are illustrative rather than actual financial losses. The threshold of 0.3 is a candidate for further evaluation, not a final production decision.

### 5.8 Compare Cost-Focused and Precision-Focused Thresholds

This section compares thresholds 0.3 and 0.8 on the validation set. Threshold 0.3 represents the lowest-cost option under the illustrative business assumptions, while threshold 0.8 represents the precision-focused candidate selected earlier.

The comparison considers precision, recall, F1-score, missed fraud cases, false alarms, and estimated cost.

```python
# Step 5.8: Compare thresholds 0.3 and 0.8 on validation data

from sklearn.metrics import precision_score, recall_score, f1_score

threshold_comparison = []

for threshold in [0.3, 0.8]:
    val_predictions = (y_val_proba_rf >= threshold).astype(int)

    missed_fraud = int(
        ((y_val == 1) & (val_predictions == 0)).sum()
    )

    false_alarms = int(
        ((y_val == 0) & (val_predictions == 1)).sum()
    )

    threshold_comparison.append({
        "Threshold": threshold,
        "Precision": precision_score(
            y_val, val_predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_val, val_predictions, zero_division=0
        ),
        "F1-score": f1_score(
            y_val, val_predictions, zero_division=0
        ),
        "Missed Fraud": missed_fraud,
        "False Alarms": false_alarms,
        "Estimated Cost": (
            missed_fraud * cost_per_missed_fraud
            + false_alarms * cost_per_false_alarm
        )
    })

display(pd.DataFrame(threshold_comparison).round(4))
```

### 5.9 Threshold Recommendation

The validation-set comparison indicates that threshold 0.3 is the preferred candidate when minimizing the estimated cost of missed fraud and false alarms under the assumptions used in this project.

At threshold 0.3, the model achieved 88.16% recall and 64.42% precision, with 9 missed fraud cases and 37 false alarms. Its estimated cost was 1,085 units. At threshold 0.8, precision increased to 92.65%, but 13 fraud cases were missed and the estimated cost increased to 1,325 units.

Therefore, threshold 0.3 is recommended for further evaluation under the assumed cost scenario. This is a provisional recommendation because the cost values are illustrative. Actual business costs and further validation would be needed before selecting a production threshold.

### 5.10 Generalization and Evaluation Limitations

```python

# 6.6 Create a Reusable Fraud Prediction Function

def predict_fraud(transaction_data, model=loaded_model, threshold=final_threshold):
    """
    Predict whether a transaction is potentially fraudulent.

    Parameters:
        transaction_data: DataFrame containing the 30 model features.
        model: Trained fraud detection pipeline.
        threshold: Probability cutoff for flagging potential fraud.

    Returns:
        Dictionary containing the fraud probability, predicted class,
        and prediction result.
    """

    # Validate the input type
    if not isinstance(transaction_data, pd.DataFrame):
        raise TypeError("Input must be a pandas DataFrame.")

    # Validate the number of features
    if transaction_data.shape[1] != 30:
        raise ValueError(
            f"Expected 30 features, but received {transaction_data.shape[1]}."
        )

    # Ensure the features are in the expected order
    expected_features = list(X_train.columns)

    if list(transaction_data.columns) != expected_features:
        raise ValueError(
            "Input columns must match the model's expected feature names "
            "and order."
        )

    # Validate the threshold
    if not 0 <= threshold <= 1:
        raise ValueError("Threshold must be between 0 and 1.")

    # Calculate the fraud probability
    probability = float(model.predict_proba(transaction_data)[0, 1])

    # Convert probability to a classification
    prediction = int(probability >= threshold)

    # Return the results
    return {
        "fraud_probability": probability,
        "predicted_class": prediction,
        "result": (
            "Potentially fraudulent transaction"
            if prediction == 1
            else "Likely legitimate transaction"
        )
    }


# Test the function using one test transaction
import pandas as pd

sample_result = predict_fraud(X_test.iloc[[0]])

print("REUSABLE FRAUD PREDICTION TEST")
print("-" * 40)
print(f"Fraud probability: {sample_result['fraud_probability']:.4f}")
print(f"Predicted class: {sample_result['predicted_class']}")
print(f"Result: {sample_result['result']}")
```
